#Funciones estadísticas

```
import numpy as np
```

```
#Lista de edades
edades=np.array([22,21,20,21,21,21,21,21,21,20,21,21,22,21])
print(edades)
edades
```

```
#Lista de estaturas
estaturas=np.array([1.79, 1.66, 1.75, 167, 1.76, 1.55, 1.60, 1.73, 1.82, 1.65, 1.61, 1.72, 1.72, 1.62])
print(estaturas)

estaturas
```

```
#Obtener medidas de tendencia central (media y mediana)y
#Medidas de dispersión(Varianza y desviacion estándar)
print("Media del array edades: ", np.mean(edades))
print("Mediana array edades: ", np.median(edades))
print("varianza del array edades: ", np.var(edades))
print("Deviacion estandar del arreglo edades: ", np.std(edades))
```
