```
import pandas as pd
import numpy as np
```

```
import matplotlib.pyplot as plt
%matplotlib inline
#Mostrar gráficas directamente en el Notebook

```

```
x= np.linspace(0,10,100)
plt.plot(x,np.cos(x))
plt.show()
```

```
x=np.linspace(0,10,100)
plt.plot(x, np.sin(x), '-', color='c', linewidth=3)
plt.xlabel("Radianes")
plt.ylabel("Valores")
plt.title("Función senoidal")
plt.suptitle("Ejemplo de matplotlib.pyplot")
plt.show()
```

```
x=np.linspace(0,10,100)
plt.plot(x, np.sin(x), '--', color='m', linewidth=3)
plt.xlabel("Radianes")
plt.ylabel("Valores")
plt.title("Función senoidal")
plt.suptitle("Ejemplo de matplotlib.pyplot")
plt.show()
```

```
x=np.linspace(0,10,100)
plt.plot(x, np.sin(x), '--', color='g', linewidth=3)
plt.xlabel("Radianes")
plt.ylabel("Valores")
plt.title("Función senoidal")
plt.grid()
plt.axis([-1,11,-1.5,1.5])
plt.suptitle("Ejemplo de matplotlib.pyplot")
plt.show()
```
